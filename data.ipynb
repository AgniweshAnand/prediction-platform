```python
import pandas as pd
import random


SOURCE_CSV = "sports_betting_predictive_analysis.csv"

df = pd.read_csv(SOURCE_CSV)

print("Dataset loaded successfully!")
print(f"Total rows: {len(df)}")
print(f"Total columns: {len(df.columns)}")

display(df.head())
```

```python
football_matches = df[
    (df["Sport"].astype(str).str.strip().str.lower() == "football") &
    (df["Actual_Winner"].astype(str).str.strip().str.lower() != "draw") &
    (df["Predicted_Winner"].astype(str).str.strip().str.lower() != "draw") &
    (df["Home_Team_Odds"].notna()) &
    (df["Away_Team_Odds"].notna())
].copy()

print(f"Valid football matches found: {len(football_matches)}")

display(football_matches.head())
```

```python
selected_match = football_matches.sample(n=1).iloc[0]

print(
    f"Selected Match: {selected_match['Match_ID']} | "
    f"{selected_match['Home_Team']} vs {selected_match['Away_Team']}"
)
```
